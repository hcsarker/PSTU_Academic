# Problem: Banker's Algorithm for Deadlock Avoidance

There are 5 processes and 3 resource types in a system.

For each process, the system knows:
1. Allocation Matrix
2. Maximum Need Matrix
3. Available Resources

Implement Banker's Algorithm to check whether the system is in a safe state or not.

If the system is safe, print the safe sequence.
If the system is unsafe, print that deadlock may occur.

```python
# ---------------------------------------------
# Banker's Algorithm for Deadlock Avoidance
# ---------------------------------------------

# Number of processes
processes = 5

# Number of resource types
resources = 3

# Allocation Matrix
# allocation[i][j] means process i currently has j type resource
allocation = [
    [0, 1, 0],   # P0
    [2, 0, 0],   # P1
    [3, 0, 2],   # P2
    [2, 1, 1],   # P3
    [0, 0, 2]    # P4
]

# Maximum Matrix
# maximum[i][j] means process i can need maximum j type resource
maximum = [
    [7, 5, 3],   # P0
    [3, 2, 2],   # P1
    [9, 0, 2],   # P2
    [2, 2, 2],   # P3
    [4, 3, 3]    # P4
]

# Available Resources
# available[j] means currently available j type resource
available = [3, 3, 2]


# Need Matrix calculation
# need = maximum - allocation
need = []

for i in range(processes):
    row = []
    for j in range(resources):
        row.append(maximum[i][j] - allocation[i][j])
    need.append(row)


def is_safe():
    # Work initially contains available resources
    work = available.copy()

    # Finish array shows whether a process has completed or not
    finish = [False] * processes

    # Safe sequence will store the order of process execution
    safe_sequence = []

    while len(safe_sequence) < processes:
        found = False

        for i in range(processes):
            # If process is not finished
            if finish[i] == False:

                # Check if need of process i is less than or equal to work
                can_execute = True

                for j in range(resources):
                    if need[i][j] > work[j]:
                        can_execute = False
                        break

                # If process can execute
                if can_execute:
                    # After execution, process releases its allocated resources
                    for j in range(resources):
                        work[j] += allocation[i][j]

                    # Mark process as finished
                    finish[i] = True

                    # Add process to safe sequence
                    safe_sequence.append(i)

                    found = True

        # If no process can execute, system is unsafe
        if found == False:
            return False, safe_sequence

    return True, safe_sequence


# Display Need Matrix
print("Need Matrix:")
for i in range(processes):
    print(f"P{i}: {need[i]}")

# Check system safety
safe, sequence = is_safe()

if safe:
    print("\nSystem is in a SAFE state.")
    print("Safe Sequence:", end=" ")

    for i in sequence:
        print(f"P{i}", end=" ")

else:
    print("\nSystem is in an UNSAFE state.")
    print("Deadlock may occur.")
```
